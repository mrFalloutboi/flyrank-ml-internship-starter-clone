## 3. Success metric

*One metric you can defend. What number means 'good'?*

The success metric I'd chose is NDCG (Normalized Discounted Cumulative Gain).
It is appropriate because it measures how well the predicted ranking matches the ideal ranking,
while giving more importance to items at the top of the list.
An NDCG score close to 1.0 means excellent ranking quality.
In practice, a score above 0.8 would be considered good performance for this task.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

ML beats a fixed rule here because ranking signals interact in complex, non-linear ways.
CTR, dwell time, and correlation values do not follow a single threshold or simple condition.
An if-statement cannot capture the messy patterns of user behavior, overlapping signals, and contextual differences.
Machine learning can learn these interactions and adapt to noisy data, making it more effective than fixed rules.
